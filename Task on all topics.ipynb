```python
#18. Create class Rectangle with private length and width. Add getter, setter and method to calculate area.

class Rectangle:
    def __init__(self, length, width):
        self.__length = length   
        self.__width = width    

    # Getters
    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    # Setters
    def set_length(self, length):
        self.__length = length

    def set_width(self, width):
        self.__width = width

 
    def area(self):
        return self.__length * self.__width


# Create object
r = Rectangle(10, 5)

# Get values
print("Length:", r.get_length())
print("Width:", r.get_width())

print("Area:", r.area())

r.set_length(12)
r.set_width(6)

print("Updated Area:", r.area())
```

```python
#20. Create multi-level inheritance: Person → Employee → Manager.


class Person:
    def __init__(self, name):
        self.name = name


class Employee(Person):
    def __init__(self, name, emp_id):
        super().__init__(name)
        self.emp_id = emp_id


class Manager(Employee):
    def __init__(self, name, emp_id, department):
        super().__init__(name, emp_id)
        self.department = department

    def display(self):
        print("Name:", self.name)
        print("Employee ID:", self.emp_id)
        print("Department:", self.department)

# Create object
m = Manager("Ravi", 101, "Sales")

# Display details
m.display()
```

```python
# 23. Create abstract class Shape with abstract method area(). Implement it in Circle and Rectangle using super().


from abc import ABC, abstractmethod


class Shape(ABC):
    def __init__(self):
        super().__init__()

    @abstractmethod
    def area(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        super().__init__()
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

class Rectangle(Shape):
    def __init__(self, length, width):
        super().__init__()
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

# Create objects
c = Circle(5)
r = Rectangle(10, 4)

# Display areas
print("Circle Area:", c.area())
print("Rectangle Area:", r.area())
```

```python
# 24. Create hybrid inheritance with classes A, B, C and D (D inherits from B and C).


class A:
    def show_A(self):
        print("Class A method")


class B(A):
    def show_B(self):
        print("Class B method")


class C(A):
    def show_C(self):
        print("Class C method")


class D(B, C):
    def show_D(self):
        print("Class D method")


d = D()

# Call methods
d.show_A()
d.show_B()
d.show_C()
d.show_D()
```

```python
# 25. Create a class Calculator that can add 2 or 3 numbers using method overloading concept
#(*args or default arguments).


class Calculator:
    def add(self, *args):
        total = 0
        for num in args:
            total += num
        return total


calc = Calculator()


print("Sum of 2 numbers:", calc.add(10, 20))


print("Sum of 3 numbers:", calc.add(10, 20, 30))
```

```python
#27. Create class Point and overload + operator to add two points.

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    
    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

  
    def display(self):
        print("Point(", self.x, ",", self.y, ")")


p1 = Point(2, 3)
p2 = Point(4, 5)


p3 = p1 + p2


print("First Point:")
p1.display()

print("Second Point:")
p2.display()

print("Sum of Points:")
p3.display()
```

```python
# 28. Create class Book and overload > operator to compare number of pages of two books.

class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    
    def __gt__(self, other):
        return self.pages > other.pages

# Create objects
book1 = Book("Python Programming", 300)
book2 = Book("Data Science", 250)

# Compare books
if book1 > book2:
    print(book1.title, "has more pages.")
else:
    print(book2.title, "has more pages.")
```

```python
# 30. Create class ComplexNumber and overload both + and - operators.
class ComplexNumber:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    
    def __add__(self, other):
        return ComplexNumber(
            self.real + other.real,
            self.imag + other.imag
        )

    
    def __sub__(self, other):
        return ComplexNumber(
            self.real - other.real,
            self.imag - other.imag
        )

    
    def display(self):
        print(f"{self.real} + {self.imag}i")

c1 = ComplexNumber(5, 3)
c2 = ComplexNumber(2, 1)

c3 = c1 + c2
c4 = c1 - c2

print("Addition:")
c3.display()

print("Subtraction:")
c4.display()
```
