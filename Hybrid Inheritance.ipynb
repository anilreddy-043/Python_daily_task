```python
## Hybrid Inheritance 
#Questions (1-10) 
#1. Create a hybrid inheritance structure with classes A → B and C → D (where D inherits from both B and C).
#Concept Hybrid Inheritance (Combination of Single + Multiple Inheritance) What to do • Create parent class A. 
#• Create class B inheriting from A.  • Create another class C. 
#• Create class D inheriting from both B and C. 
#How to solve 
#1. Create class A.  2. Add one method. 
#3. Create class B(A). 
#4. Create class C.  
#5. Create class D(B, C). 
#6. Create an object of D. 
#7. Call methods from all parent classes.  

class parentA:
    def method_parentA(self):
        return "This is method from class A"

class parentB(parentA):
    def method_parentB(self):
        return "This is method from class B"

class parentC:
    def method_parentC(self):
        return "This is method from class B"

class parentD(parentB,parentC):
    def method_parentD(self):
        return "This is method from class B"



obj = parentD()

obj.method_parentA() 
obj.method_parentB()   
obj.method_parentC()   
obj.method_parentD()  
```

```python
#2. Create Person, Employee, Student, and TeachingAssistant using hybrid inheritance. 
#Concept Real-world Hybrid Inheritance What to do 
#• Person is the base class.  
#• Employee inherits Person.  
#• Student inherits Person.  
#• TeachingAssistant inherits Employee and Student. 
#Steps
#1. Store name in Person.  2. Employee stores salary.  3. Student stores marks.  4. TeachingAssistant stores department.  5. Display all information.



# Base class
class Person:
    def __init__(self, name):
        self.name = name

# Employee inherits from Person
class Employee(Person):
    def __init__(self, name, salary):
        Person.__init__(self, name)
        self.salary = salary

# Student inherits from Person
class Student(Person):
    def __init__(self, name, marks):
        Person.__init__(self, name)
        self.marks = marks

# TeachingAssistant inherits from Employee and Student
class TeachingAssistant(Employee, Student):
    def __init__(self, name, salary, marks, department):
        Employee.__init__(self, name, salary)
        Student.__init__(self, name, marks)
        self.department = department
    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)
        print("Marks:", self.marks)
        print("Department:", self.department)

# Create object
ta = TeachingAssistant("Rahul", 30000, 85, "Computer Science")

# Display information
ta.display()
```

```python
#3. Create Vehicle → Car, Electric → ElectricCar (ElectricCar inherits from both).
#Concept Hybrid Inheritance What to do 
#• Vehicle is parent.  
#• Car inherits Vehicle.  
#• Electric is another parent.  
#• ElectricCar inherits Car and Electric.  
#Steps 1. Vehicle has start().  2. Car has drive().  3. Electric has battery().  4. ElectricCar accesses all methods.


class Vehicle:
    def start(self):
        print("Vehicle started")


class Car(Vehicle):
    def drive(self):
        print("Car is driving")


class Electric:
    def battery(self):
        print("Battery is fully charged")


class ElectricCar(Car, Electric):
    pass

# Create object
ec = ElectricCar()


ec.start()
ec.drive()
ec.battery()
        
```

```python
#4. Demonstrate Hybrid Inheritance and print Method Resolution Order (MRO). 
#Concept Method Resolution Order (MRO) 
#How to solve 1. Create four classes. 
#2. Give same method name in multiple classes.  
#3. Print ClassName.mro(). 
#4. Observe Python's searching order. 


class A:
    def show(self):
        print("Method from Class A")

class B(A):
    def show(self):
        print("Method from Class B")

class C(A):
    def show(self):
        print("Method from Class C")

class D(B, C):
    pass

# Create object of D
obj = D()

# Call the method

obj.show()

# Print Method Resolution Order
print("\nMethod Resolution Order (MRO):")
print(D.mro())

```
