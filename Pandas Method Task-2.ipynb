```python
#Q30 Real Industry Mini Project
"""Your manager asks you to prepare a cleaned dataset for the Machine Learning team.
Perform all of these steps:
1. Load the dataset.
2. Display dataset information.
3. Check missing values.
4. Remove duplicate rows.
5. Rename the following columns:
o CustomerId → Customer_ID
o CreditScore → Credit_Score
o EstimatedSalary → Salary
o Exited → Target
6. Convert:
o Age → float
o Balance → float
o Target → string
7. Replace missing values:
o Age → median
o Balance → mean
8. Remove:
o RowNumber
o Surname
9. Display:
o First 10 rows
o Statistical summary
o Value counts of Target
o Final dataset information"""

# 1. Load the dataset.
df = pd.read_csv("C:/Users/anilr/Downloads/Bank_customer_churn.csv")
print(df)

# 2. Display dataset information.
df.info()
print(df)

# 3. Check missing values.
df.isnull().sum()
print(df)
# 4. Remove duplicate rows.
df.duplicated().sum()
print(df)

# 5. Rename the following columns:
#o CustomerId → Customer_ID
#o CreditScore → Credit_Score
#o EstimatedSalary → Salary
#o Exited → Target
df.rename(columns = {"CustomerId":"Customer_ID",
                          "CreditScore":"Credit_Score",
                          "EstimatedSalary":"Salary",
                          "Exited":"Target",
                         },inplace = True)
print(df)

#6. Convert:
#o Age → float
df["Age"] = df["Age"].astype(float)
print(df["Age"])
#o Balance → float
df["Balance"] = df["Balance"].astype(float)
print(df["Balance"])
#o Target → string
df["Target"] = df["Target"].astype(float)
print(df["Target"])

# 7. Replace missing values:
#o Age → median
Age_median =  df["Age"].median()
print(Age_median)
#o Balance → mean
Balance_median =  df["Balance"].mean()
print(Balance_median)

#8. Remove:
#o RowNumber
df.drop("RowNumber", axis=1, inplace=True)
print(df)
#o Surname
df.drop("Surname", axis=1, inplace=True)
print(df)

#9. Display:
#o First 10 rows
df.head(10)
print(df)
#o Statistical summary
df.head(10)
print(df)
#o Value counts of Target
df["Target"].value_counts()
print(df)
#o Final dataset information
df.info()
print(df)

```
