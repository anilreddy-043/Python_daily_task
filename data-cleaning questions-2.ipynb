# Data Type & Invalid Values

```python
# 11.The Age column contains numbers as well as values like "unknown" and
#"twenty five". Clean this column and convert it into a proper numeric column.
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Age"] = df["Age"].fillna(df["Age"].median())
abs(df["Age"].unique())
```

```python
df.columns
```

```python
# 12.The Annual Income column contains values such as 650000 INR, ₹720000,
# 1.2L, and N/A. Clean the column and convert the valid values into numericformat.
df["Annual Income"].nunique()
df["Annual Income"] = pd.to_numeric(df["Annual Income"], errors = "coerce")
df['Annual Income'] = df["Annual Income"] = df["Annual Income"].fillna(df["Annual Income"].median())
df["Annual Income"].unique()
df["Annual Income"].isnull().sum()
```

```python
# 13.Clean the Credit Score column and identify values that cannot be converted into numbers.
df["Credit Score"].isna()
df["Credit Score"] = pd.to_numeric(df["Credit Score"], errors = "coerce")
df["Credit Score"] = df["Credit Score"].fillna(df["Credit Score"].median())
abs(df["Credit Score"].isnull().sum())
```

```python
# 14.The Loan Amount column contains currency symbols, text, and numeric
# values. Remove the unwanted characters and convert it into numeric format.
df["Loan Amount (Inr)"].isna().sum()
df["Loan Amount (Inr)"] = pd.to_numeric(df["Loan Amount (Inr)"], errors = "coerce")
df["Loan Amount (Inr)"] = df["Loan Amount (Inr)"].fillna(df["Loan Amount (Inr)"].median())
abs(df["Loan Amount (Inr)"].isna().sum())
df["Loan Amount (Inr)"].unique
```

```python
#  15.The Monthly Visits column contains values such as "five", "N/A", and "-".
# Convert valid values into numeric values and handle invalid values.
df["Monthly Visits"].nunique()
df["Monthly Visits"] = pd.to_numeric(df["Monthly Visits"], errors = "coerce")
df["Monthly Visits"] = df["Monthly Visits"].fillna(df["Monthly Visits"].median())
abs(df["Monthly Visits"].unique())
```

```python
# 16.Clean the Debt to Income column where some values are written like 35%
# and 0.4x. Convert the valid values into a consistent numeric format.
df["Debt_To_Income"].nunique()
df["Debt_To_Income"] = pd.to_numeric(df["Debt_To_Income"], errors = "coerce")
df["Debt_To_Income"] = df["Debt_To_Income"].fillna(df["Debt_To_Income"].median())
abs(df["Debt_To_Income"].unique())
```

```python
#17.After cleaning the numerical columns, check which columns are still stored as object data type.
df.dtypes.value_counts()
```

# Categorical Cleaning

```python
# 18.Find all unique values in the Employment-Type column.
df["Employment-Type"].unique()
```

```python
# 19.Standardize the Employment-Type column so that values such as salaried, SALARIED, and Salaried become one consistent value.
df["Employment-Type"] = df["Employment-Type"].replace({"Self employed": "Self-Employed",
                                                       "self-employed": "Self-Employed",
                                                       "SALARIED": "Salaried",
                                                       "salaried": "Salaried",
                                                       "Salaried": "Salaried",
                                                       "Unknown": "nan",
                                                       "nan": "nan"
})

df["Employment-Type"].unique()
```

```python
# 20.Clean the Education Level column and make values such as graduate, GRADUATE, and Graduate consistent.
df["Education  Level"] = df["Education  Level"].replace({"postgraduate": "Post Graduate",
                                                       "GRADUATE": "Graduate",
                                                       "Postgraduate": "Post Graduate",
                                                         "graduate":"Graduate"
})

df["Education  Level"].unique()
```

```python
# 21.Clean the City column. Identify incorrect or inconsistent values such as hyderbad, Hyd, Bangalore, and Bengaluru.
df["City"] = df["City"].replace({"MUMBAI": "Mumbai",
                                                         "mumbai":"Mumbai",
                                                       "hyderbad": "Hyderabad",
                                                       "Bangalore": "Bengaluru",
                                                         "Hyd":"Hyderabad",
                                                        })
df["City"].unique()
```

```python
# 22.Find all unique values in the Customer Segment column and identify invalid categories.

df["Customer_Segment"] = df["Customer_Segment"].replace({"mass": "Mass",
                                                         "PREMIUM":"Premium",
                                                       "premium": "Premium",
                                                        })
df['Customer_Segment'].unique()
```

```python
# 23.Clean the Loan Type column so that values such as personal, PERSONAL, and personal loan are standardized.
df["Loan_Type"] = df["Loan_Type"].replace({"PERSONAL": "Personal",
                                            "auto":"Auto",
                                            "Home": "Home Loan",
                                                        })
df['Loan_Type'].unique()
```

```python
df.columns
```

```python
# 24.Clean the Loan Approved column where values include Yes, yes, Y, 1, No, NO, N, and 0. Convert them into only Yes/No.

df["Loan Approved?"] = df["Loan Approved?"].replace({"NO": "No",
                                            "yes":"Yes",
                                            "0": "No",
                                            "1":"Yes",
                                            "Y":"Yes",
                                            "N":"No"
                                                        })
df['Loan Approved?'].unique()
```

# Outliers & Data Validation

```python
# 25.Check the Age column and identify unrealistic ages such as negative ages or ages above 100.


df['Age'] = abs(df['Age'].replace({130:99}))
df["Age"].unique()
```

```python
# 26.Check the Credit Score column and identify values outside the valid credit- score range.
df['Credit Score'].unique()
numeric_scores = pd.to_numeric(df["Credit Score"], errors="coerce")
invalid_df = df[(numeric_scores < 300) | (numeric_scores > 900) | df["Credit Score"].isna() | numeric_scores.isna()]
invalid_df
```

```python
df.columns
```

```python
# 27.Detect outliers in Annual Income using the IQR method.

Q1 = df["Annual Income"].quantile(0.25)
Q3 = df["Annual Income"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
lower_bound
upper_bound = Q3 + 1.5 * IQR
upper_bound

outliers_df = df[(df["Annual Income"] < lower_bound) | (df["Annual Income"] > upper_bound)]
outliers_df


```

```python
# 28.Detect outliers in Loan Amount using the IQR method.
Q1 = df["Loan Amount (Inr)"].quantile(0.25)
Q3 = df["Loan Amount (Inr)"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Loan Amount (Inr)"] < lower_bound) |
    (df["Loan Amount (Inr)"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))
print(outliers)
```

```python
# 29.Check the Debt to Income column and identify impossible values such as negative values or values greater than 1.
print(df.loc[
    (df["Debt_To_Income"] < 0) |
    (df["Debt_To_Income"] > 1),
    "Debt_To_Income"
])
```
