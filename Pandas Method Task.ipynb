```python
# 18. Create abstract class Shape with abstract method area(). Implement it in Circle and Square using super().
from abc import ABC, abstractmethod
import math

# Abstract class
class Shape(ABC):
    def __init__(self, value):
        self.value = value
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        super().__init__(radius)   # Call Shape constructor

    def area(self):
        return math.pi * self.value ** 2


# Square class
class Square(Shape):
    def __init__(self, side):
        super().__init__(side)     # Call Shape constructor

    def area(self):
        return self.value ** 2


# Creating objects
c = Circle(7)
s = Square(5)

print("Circle Area:", round(c.area(), 2))
print("Square Area:", s.area())
```

```python
#  21. Create class Point and overload + and > operators.
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overload + operator
    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    # Overload > operator
    def __gt__(self, other):
        return (self.x + self.y) > (other.x + other.y)

    def display(self):
        print(f"({self.x}, {self.y})")


# Creating objects
p1 = Point(2, 3)
p2 = Point(4, 5)

# Using + operator
p3 = p1 + p2
print("After adding points:")
p3.display()

# Using > operator
if p1 > p2:
    print("p1 is greater than p2")
else:
    print("p2 is greater than or equal to p1")
```

```python
# 26. Create a class Student with private marks list.
"""• Setter should validate each mark (0-100)
• Overload > operator to compare two students based on average
• Raise custom exception if marks list is empty"""
# Custom Exception
class EmptyMarksError(Exception):
    pass


class Student:
    def __init__(self, marks):
        self.__marks = []
        self.set_marks(marks)

    # Setter with validation
    def set_marks(self, marks):
        if len(marks) == 0:
            raise EmptyMarksError("Marks list cannot be empty.")

        for mark in marks:
            if mark < 0 or mark > 100:
                raise ValueError("Each mark must be between 0 and 100.")

        self.__marks = marks

    # Getter
    def get_marks(self):
        return self.__marks

    # Calculate average
    def average(self):
        return sum(self.__marks) / len(self.__marks)

    # Overload > operator
    def __gt__(self, other):
        return self.average() > other.average()


# Main Program
try:
    s1 = Student([80, 90, 85])
    s2 = Student([70, 75, 80])

    print("Student 1 Average:", s1.average())
    print("Student 2 Average:", s2.average())

    if s1 > s2:
        print("Student 1 has higher average.")
    else:
        print("Student 2 has higher or equal average.")

    # Uncomment to test exceptions
    # s3 = Student([])          # EmptyMarksError
    # s4 = Student([90, 105])   # ValueError

except EmptyMarksError as e:
    print("Error:", e)

except ValueError as e:
    print("Error:", e)
```

```python
# 27. Write a menu-driven banking system using OOP + File Handling. Store account details in a file. Handle FileNotFoundError, ValueError, and custom exceptions.
# Menu-Driven Banking System using OOP + File Handling

# Custom Exception
class InsufficientBalanceError(Exception):
    pass


class BankAccount:
    FILE_NAME = "accounts.txt"

    def __init__(self, acc_no, name, balance=0):
        self.acc_no = acc_no
        self.name = name
        self.balance = balance

    # Save account details
    def save_account(self):
        with open(self.FILE_NAME, "a") as file:
            file.write(f"{self.acc_no},{self.name},{self.balance}\n")

    # Deposit amount
    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")
        self.balance += amount
        print("Deposit Successful.")
        self.update_file()

    # Withdraw amount
    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")
        if amount > self.balance:
            raise InsufficientBalanceError("Insufficient Balance.")
        self.balance -= amount
        print("Withdrawal Successful.")
        self.update_file()

    # Update account in file
    def update_file(self):
        try:
            with open(self.FILE_NAME, "r") as file:
                accounts = file.readlines()

            with open(self.FILE_NAME, "w") as file:
                for acc in accounts:
                    data = acc.strip().split(",")
                    if data[0] == self.acc_no:
                        file.write(f"{self.acc_no},{self.name},{self.balance}\n")
                    else:
                        file.write(acc)

        except FileNotFoundError:
            print("Error: Account file not found.")

    # Display account
    def display(self):
        print("\nAccount Number:", self.acc_no)
        print("Name:", self.name)
        print("Balance:", self.balance)


# Search account
def search_account(acc_no):
    try:
        with open("accounts.txt", "r") as file:
            for line in file:
                data = line.strip().split(",")
                if data[0] == acc_no:
                    return BankAccount(data[0], data[1], float(data[2]))
    except FileNotFoundError:
        print("Error: Account file not found.")
    return None


# Menu
while True:
    print("\n===== BANK MENU =====")
    print("1. Create Account")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Display Account")
    print("5. Exit")

    choice = input("Enter Choice: ")

    try:
        if choice == "1":
            acc = input("Account Number: ")
            name = input("Name: ")
            balance = float(input("Initial Balance: "))
            account = BankAccount(acc, name, balance)
            account.save_account()
            print("Account Created Successfully.")

        elif choice == "2":
            acc = input("Account Number: ")
            account = search_account(acc)
            if account:
                amount = float(input("Deposit Amount: "))
                account.deposit(amount)
            else:
                print("Account Not Found.")

        elif choice == "3":
            acc = input("Account Number: ")
            account = search_account(acc)
            if account:
                amount = float(input("Withdraw Amount: "))
                account.withdraw(amount)
            else:
                print("Account Not Found.")

        elif choice == "4":
            acc = input("Account Number: ")
            account = search_account(acc)
            if account:
                account.display()
            else:
                print("Account Not Found.")

        elif choice == "5":
            print("Thank You!")
            break

        else:
            print("Invalid Choice.")

    except ValueError as e:
        print("ValueError:", e)

    except InsufficientBalanceError as e:
        print("Error:", e)

    except FileNotFoundError:
        print("Error: File not found.")
```

```python
# 28. Using dictionary + exception handling, create a student management system that
"""can:
• Add student
• Update marks
• Delete student
• Display all Raise custom exception if student not found."""
# Custom Exception
class StudentNotFoundError(Exception):
    pass


# Dictionary to store student details
students = {}

while True:
    print("\n===== STUDENT MANAGEMENT SYSTEM =====")
    print("1. Add Student")
    print("2. Update Marks")
    print("3. Delete Student")
    print("4. Display All Students")
    print("5. Exit")

    choice = input("Enter your choice (1-5): ")

    try:
        if choice == "1":
            name = input("Enter student name: ")
            marks = int(input("Enter marks: "))

            if marks < 0 or marks > 100:
                raise ValueError("Marks must be between 0 and 100.")

            students[name] = marks
            print("Student added successfully.")

        elif choice == "2":
            name = input("Enter student name to update: ")

            if name not in students:
                raise StudentNotFoundError("Student not found.")

            marks = int(input("Enter new marks: "))

            if marks < 0 or marks > 100:
                raise ValueError("Marks must be between 0 and 100.")

            students[name] = marks
            print("Marks updated successfully.")

        elif choice == "3":
            name = input("Enter student name to delete: ")

            if name not in students:
                raise StudentNotFoundError("Student not found.")

            del students[name]
            print("Student deleted successfully.")

        elif choice == "4":
            if not students:
                print("No student records found.")
            else:
                print("\nStudent Records")
                for name, marks in students.items():
                    print(f"{name} : {marks}")

        elif choice == "5":
            print("Exiting program...")
            break

        else:
            print("Invalid choice.")

    except ValueError as e:
        print("ValueError:", e)

    except StudentNotFoundError as e:
        print("Error:", e)
```

```python
# 29. Create a class Employee with private salary.
"""• Setter should not allow more than 15% increase at once
• Keep history of all salary changes
• Add static method to calculate tax
• Overload > operator to compare employees by salary"""
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary
        self.__history = [salary]   # Store salary history

    # Getter
    def get_salary(self):
        return self.__salary

    # Setter (maximum 15% increase allowed)
    def set_salary(self, new_salary):
        if new_salary > self.__salary * 1.15:
            print("Salary increase cannot exceed 15% at a time.")
        else:
            self.__salary = new_salary
            self.__history.append(new_salary)
            print("Salary updated successfully.")

    # Display salary history
    def show_history(self):
        print("Salary History:", self.__history)

    # Static method to calculate tax (10%)
    @staticmethod
    def calculate_tax(salary):
        return salary * 0.10

    # Overload > operator
    def __gt__(self, other):
        return self.__salary > other.__salary


# Creating objects
e1 = Employee("Anil", 50000)
e2 = Employee("Ravi", 60000)

# Valid salary update
e1.set_salary(55000)

# Invalid salary update (more than 15%)
e1.set_salary(70000)

print("\nCurrent Salary:", e1.get_salary())
e1.show_history()

print("Tax on Salary:", Employee.calculate_tax(e1.get_salary()))

# Compare salaries
if e1 > e2:
    print("Anil has a higher salary.")
else:
    print("Ravi has a higher or equal salary.")
```

```python
# 30. Design a complete system combining:
"""• Abstract class Vehicle
• Multiple inheritance (Electric + Fuel)
• Operator overloading (compare by speed)
• Private attributes + getter/setter
• Custom exceptions
• File handling to save vehicle details
• Duck Typing for start() method"""
from abc import ABC, abstractmethod

# ---------------- Custom Exception ----------------
class InvalidSpeedError(Exception):
    pass


# ---------------- Abstract Class ----------------
class Vehicle(ABC):
    def __init__(self, name, speed):
        self.__speed = 0
        self.name = name
        self.set_speed(speed)

    # Getter
    def get_speed(self):
        return self.__speed

    # Setter
    def set_speed(self, speed):
        if speed < 0:
            raise InvalidSpeedError("Speed cannot be negative.")
        self.__speed = speed

    @abstractmethod
    def vehicle_type(self):
        pass

    # Operator Overloading (>)
    def __gt__(self, other):
        return self.__speed > other.__speed


# ---------------- Parent Classes ----------------
class Electric:
    def charge(self):
        print("Charging electric vehicle...")


class Fuel:
    def fill_fuel(self):
        print("Filling fuel tank...")


# ---------------- Child Class (Multiple Inheritance) ----------------
class HybridVehicle(Vehicle, Electric, Fuel):
    def __init__(self, name, speed):
        super().__init__(name, speed)

    def vehicle_type(self):
        print("Hybrid Vehicle")

    # Duck Typing Method
    def start(self):
        print(self.name, "started.")


# ---------------- File Handling ----------------
def save_vehicle(vehicle):
    with open("vehicles.txt", "a") as file:
        file.write(f"{vehicle.name},{vehicle.get_speed()}\n")


# ---------------- Duck Typing Function ----------------
def start_vehicle(obj):
    obj.start()      # Any object having start() method works


# ---------------- Main Program ----------------
try:
    v1 = HybridVehicle("Tesla Hybrid", 180)
    v2 = HybridVehicle("Toyota Hybrid", 150)

    v1.vehicle_type()
    v1.charge()
    v1.fill_fuel()

    # Operator Overloading
    if v1 > v2:
        print(v1.name, "is faster than", v2.name)
    else:
        print(v2.name, "is faster")

    # Duck Typing
    start_vehicle(v1)

    # Save to File
    save_vehicle(v1)
    save_vehicle(v2)

    print("Vehicle details saved successfully.")

except InvalidSpeedError as e:
    print("Error:", e)

except FileNotFoundError:
    print("File not found.")

except Exception as e:
    print("Unexpected Error:", e)
```
