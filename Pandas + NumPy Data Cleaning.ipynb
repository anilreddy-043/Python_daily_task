# Pandas + NumPy Data Cleaning

```python
# 1. Load the CSV using Pandas and display the first 10 rows.
import pandas as pd
df = pd.read_csv("C:\\Users\\anilr\\Downloads\\customer_data - customer_data.csv")
df.head(10)
```

```python
# 2. Check the number of rows and columns in the dataset.
df.shape
```

```python
# 3. Display all column names and identify which column names contain  leading/trailing spaces or inconsistent capitalization.  
for i in df.columns:
    print(i)

    if i != i.title():
        df = df.rename(columns = {i:i.title()})
        print(df)
    

```

```python
# 4. Check the data type of every column using Pandas. 
df.dtypes
```

```python
# 5 calculate the number and persentage of missing valuse in every column.uyy

(df.isna().sum() / len(df)) * 100
```

```python
# 6. Find the rows where Customer ID is missing. 
df[df["Customer ID"].isna()]
```

```python
# 7. Remove leading/trailing whitespace from all column names and convert them  to lowercase with underscores.  
df.columns.str.replace(" ","_",regex = True)
```

```python
# 8. Identify duplicate Customer ID values and count how many duplicates exist. 
df["Customer ID"].duplicated().sum()
df[df["Customer ID"].duplicated()]
```

```python
# 9. Remove duplicate customer records while keeping the first occurrence.  
df = df.drop_duplicates(keep = "first")
df
```

```python
# 10.Find all unique values in the customer_segment column and identify  inconsistent categories such as Consumer, consumer, consumer , etc.  
df["customer_segment"].value_counts()
df.rename(columns = {"Consumer":"consumer","SMB": "SMB."},inplace = True)
df["customer_segment"].value_counts()
```

```python
# 11.Clean the customer_segment column by standardizing capitalization and  whitespace. 

df["customer_segment"] = df["customer_segment"].str.strip().str.title()
df.head()
#df["customer_segment"].unique()
```

```python
# 12.Find all unique values in the gender column and convert them into consistent  categories such as Male and Female.  
df["gender"].unique()
df["gender"] = df["gender"].str.strip().str.title()
df["gender"].unique()
df["gender"] = df["gender"].str.strip().str.upper().replace({
    "M": "Male",
    "F": "Female"
})
df["gender"]
df["gender"].unique()
```

```python
# 13.Clean the is_active column so values such as Yes, yes, Y, 1 become 1, and  No, no, N, 0 become 0. 
df["is_active"].unique()
df["is_active"] = df["is_active"].str.strip().str.upper().replace({
    "yes": "Yes",
    "Y":"Yes",
    "no": "NO",
    "1":"Yes",
    "0":"No",
    "N":"nan",
    "NAN":"nan"
    
})
df["is_active"].unique()
```

```python
# 14.Convert the churn_flag column from categorical values (Yes/No) into a binary  numerical column (1/0). 
df["churn_flag"] = df["churn_flag"].map({"Yes": 1, "No": 0})

df
```

```python
# 15.Identify all values in income_usd that cannot be directly interpreted as  numbers.  
df.loc[pd.to_numeric(df['income_usd'], errors='coerce').isna(), 'income_usd'].unique()
```
