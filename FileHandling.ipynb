```python
# 26. Create a program that keeps a record of daily expenses in a file and can show total expenses.

filename = "expenses.txt"

while True:
    print("\n1. Add Expense")
    print("2. Show Total Expenses")
    print("3. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        amount = float(input("Enter expense amount: "))
        with open(filename, "a") as file:
            file.write(str(amount) + "\n")
        print("Expense added successfully.")

    elif choice == "2":
        total = 0
        try:
            with open(filename, "r") as file:
                for line in file:
                    total += float(line.strip())
            print("Total Expenses:", total)
        except FileNotFoundError:
            print("No expense record found.")

    elif choice == "3":
        print("Exiting...")
        break

    else:
        print("Invalid choice. Please try again.")
```

```python
# 27. Write a program to remove all blank lines from a file.

input_file = input()
output_file = input()

try:
    with open(input_file, "r") as infile, open(output_file, "w") as outfile:
        for line in infile:
            if line.strip():      # Skip blank lines
                outfile.write(line)

    print("Blank lines removed successfully.")

except FileNotFoundError:
    print("Error: Input file not found.")
```

```python
# 28. Read a file and write only the unique lines into another file (remove duplicates).
# Program to remove duplicate lines from a file

input_file = input("Enter input file name: ")
output_file = input("Enter output file name: ")

try:
    with open(input_file, "r") as infile:
        lines = infile.readlines()

    unique_lines = []

    for line in lines:
        if line not in unique_lines:
            unique_lines.append(line)

    with open(output_file, "w") as outfile:
        outfile.writelines(unique_lines)

    print("Duplicate lines removed successfully.")

except FileNotFoundError:
    print("Error: Input file not found.")
```

```python
# 29. Create a menu-driven program:
'''• Write to file
• Read file
• Append to file
• Exit'''

# Menu-driven program for file operations

filename = "data.txt"

while True:
    print("\n--- File Menu ---")
    print("1. Write to File")
    print("2. Read File")
    print("3. Append to File")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        data = input("Enter text to write: ")
        with open(filename, "w") as file:
            file.write(data)
        print("Data written successfully.")

    elif choice == "2":
        try:
            with open(filename, "r") as file:
                print("\nFile Contents:")
                print(file.read())
        except FileNotFoundError:
            print("File not found.")

    elif choice == "3":
        data = input("Enter text to append: ")
        with open(filename, "a") as file:
            file.write("\n" + data)
        print("Data appended successfully.")

    elif choice == "4":
        print("Exiting program...")
        break

    else:
        print("Invalid choice! Please try again.")
```

```python
# 30. Write a program that handles file exceptions (FileNotFoundError, PermissionError) properly while reading and writing files.
# Program to handle file exceptions while reading and writing files

input_file = input("Enter input file name: ")
output_file = input("Enter output file name: ")

try:
    # Read from the input file
    with open(input_file, "r") as infile:
        data = infile.read()

    # Write to the output file
    with open(output_file, "w") as outfile:
        outfile.write(data)

    print("File copied successfully.")

except FileNotFoundError:
    print("Error: Input file not found.")

except PermissionError:
    print("Error: Permission denied while accessing the file.")

except Exception as e:
    print("An unexpected error occurred:", e)
```
