# Level 5: Advanced Iterator Problems

```python
# 21 Create a custom iterator that prints the first 15 natural numbers.
class natural:
    def __init__(self):
        self.num = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.num > 15:
            raise StopIteration

        value = self.num
        self.num += 1
        return value

M = natural()

for i in M:
    print(i)
```

```python
# 22 Create a custom iterator that prints all multiples of 3 up to 30.
class MultiplesOfThree:
    def __init__(self):
        self.num = 3

    def __iter__(self):
        return self

    def __next__(self):
        if self.num > 30:
            raise StopIteration

        value = self.num
        self.num += 3
        return value

M = MultiplesOfThree()

for i in M:
    print(i)
```

```python
# 23 Create a custom iterator that prints all alphabet letters from A to Z.
class Alphabet:
    def __init__(self):
        self.letter = 'A'

    def __iter__(self):
        return self

    def __next__(self):
        if self.letter > 'Z':
            raise StopIteration

        current = self.letter
        self.letter = chr(ord(self.letter) + 1)
        return current

A = Alphabet()

for i in A:
    print(i)
```

```python
# 24 Create a custom iterator that prints all vowels from the string below.
#text = "Artificial Intelligence"
class VowelIterator:
    def __init__(self, text):
        self.text = text
        self.index = 0
    def __iter__(self):
        return self
    def __next__(self):
        vowels = "AEIOUaeiou"
        
        while self.index < len(self.text):
            char = self.text[self.index]
            self.index += 1

            if char in vowels:
                return char

        raise StopIteration

text = "Python Generator Iterator"

V = VowelIterator(text)

for ch in V:
    print(ch)
            
            
```

```python
# 25 Create a custom iterator that prints the Fibonacci sequence up to 10 terms.
class Fibonacci:
    def __init__(self):
        self.a = 0
        self.b = 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= 10:
            raise StopIteration

        value = self.a
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return value

F = Fibonacci()

for num in F:
    print(num)
```

# Level 6: Advanced Generator Problems

```python
# 26 Create a generator that yields the Fibonacci sequence up to 15 terms.
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b


for num in fibonacci(15):
    print(num)
    
```

```python
# 27 Create a generator that yields all prime numbers between 1 and 100.

def prime_generator(n):
    count = 0
    num = 2
    while count < n:
        is_prime = True
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            yield num
            count += 1

        num += 1


for prime in prime_generator(20):
    print(prime)
```

```python
# 28 Create a generator that yields all factors of a given number.
#Example Input: num = 24
def factors(num):
    for i in range(1, num + 1):
        if num % i == 0:
            yield i


num = 24


for factor in factors(num):
    print(factor)
```

```python
# 30 Create an infinite generator that yields multiples of 5. Print only the first 15 values using next().

def multiples_of_five():
    num = 5
    while True:
        yield num
        num += 5


gen = multiples_of_five()

for _ in range(15):
    print(next(gen))
```
