Lambda funtions

```python
# 1 Write a lambda function to return the square of a number. Example: Input: 5 Output: 25 
s = lambda x :x*x
s(2)
```

```python
# 2. Write a lambda function to return the cube of a number. Example: Input: 4 
s = lambda x : x*x*x
s(4)
```

```python
# 3. Write a lambda function to add two numbers. Example: Input: 10, 20 Output: 30 
add = lambda a,b : a+b
print(add(10,20))
```

```python
# 4. Write a lambda function to find the maximum of two numbers. Example: Input: 15, 8 Output: 15 
max1 = lambda a,b : max(a,b)
print(max(15,8))
```

```python
#5 Write a lambda function that returns "Even" if a number is even; otherwise "Odd". Input: 9 Output: Odd 
check_odd = lambda a : "odd" if a % 2 !=0 else "even"
check_odd(9)
```

```python
#6. Use map() with lambda to square every element of a list. Input: [1,2,3,4]  Output: [1,4,9,16] 
num = [1,2,3,4]
result = list(map(lambda x: x*x, num)) 
print(result)
```

```python
# 7. Use filter() with lambda to keep only even numbers. Input: [1,2,3,4,5,6]  Output: [2,4,6] 

numbers=[1,2,3,4,5,6]

result = list(filter(lambda x: x % 2 == 0, numbers)) 
print(result)
```

```python
#8Use filter() with lambda to keep numbers greater than 50. Input: [20,45,55,70]  Output: [55,70] 
num = [20,45,55,70]
res = list (filter(lambda x : x>50 ,num))
print(res)
```

```python
# 9.Sort a list of numbers in ascending order using lambda. [9,5,1,8,2] 
numbers = [9, 5, 1, 8, 2]

sorted_numbers = sorted(numbers, key=lambda x: x)

print(sorted_numbers)
```
