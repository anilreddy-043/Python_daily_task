```python
#2. Print all even numbers from 1 to 50 using for loop and skip numbers divisible by 10 using continue.
for i in range(1,51):
    if i % 2 == 0:
        if i % 10==0:
            continue
        print(i,end= "  ")
    
        
```

```python
#3. Take numbers from user until they enter 0. Calculate sum using while loop.
sum = 0

num = int(input())

while num != 0:
    sum += num
    num = int(input())

print(sum)
```

```python
## 7. Take a string and perform: upper(), lower(), title(), strip(), replace(), find() and count().

st = input()

print(st.upper())
print(st.lower())
print(st.title())
print(st.strip())
print(st.count("st"))
print(st.find("a"))
print(st.replace("a","@"))

```

```python
#8. Create a list of 5 numbers. Use append(), insert(), remove(), pop(), sort() and reverse().

list1 = [1,410,1.22,789,56]
list1.append(1010)

print(list1)

list1.insert(0,5)
print(list1)

list1.remove(5)
print(list1)

list1.pop()
print(list1)

list1.sort()
print(list1)

list1.reverse()
print(list1)
```

```python
# 9. Create a tuple and use count() and index() methods. Then convert it into a list.
tuple1 = (1,5,7,6,7)
co = tuple1.count(7)
print(co)

coo = tuple1.index(5)
print(coo)

li = list(tuple1)
print(li)
```
