```python
#3. (Hard) Take two numbers start and end from the user. 
#Using while loop, print all Armstrong numbers between start and end.
#(Example: 153 is Armstrong because 1³ + 5³ + 3³ = 153) 

start = int(input("Enter start: "))
end = int(input("Enter end: "))

num = start

while num <= end:
    temp = num
    digits = len(str(num))
    total = 0

    while temp > 0:
        digit = temp % 10
        total = total + digit ** digits
        temp = temp // 10

    if total == num:
        print(num)

    num += 1
```

```python
# 7. Reverse a Number Question Write a program to reverse a number using a while loop.
# Explanation Example: 1234 → 4321 Steps: • Take last digit using % 10 
#• Add digit into reverse number  • Remove last digit using // 10 


num = int(input())
rev = 0

while num > 0 :
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10
print(rev)
    
 
```
