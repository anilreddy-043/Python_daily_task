```python
#1. Create a Calculator class that can add 2 numbers or 3 numbers using the same method name.
#-->Concept
#Method Overloading (Python uses default arguments or *args to achieve overloading.)
#-->What to do
#• Create a class Calculator.
#• Create one method add().
#• The method should work for either 2 numbers or 3 numbers.
#-->How to solve
#1. Create class Calculator.
#2. Define method add().
#3. Use default arguments or *args.
#4. Check how many numbers are provided.
#5. Return the sum.

class Calculator:
    def add(self,a,b,c=0,d=0,*args):
        return a+b+c+d

A = Calculator()
A.add(3,4,5,5)
```

```python
#2. Create an Area class that calculates the area of a square and rectangle using the same method.
"""Concept
Method Overloading using default arguments.
What to do
• One parameter → Square.
• Two parameters → Rectangle.
How to solve
1. Create class Area.
2. Define method calculate().
3. If only one value is provided, calculate the square's area.
4. If two values are provided, calculate the rectangle's area.
5. Return the result."""

class Area:
    def calculate(self,l,b=0):
        if b is 0:
            return l * l
        else:
            return l * b
        
A = Area()

print(A.calculate(5))   
```

```python
##### 4. Create a MathOperations class that multiplies any number of values.
'''Concept
Method Overloading using *args
What to do
Multiply:
• 2 numbers
• 3 numbers
• 4 numbers
• More numbers
How to solve
1. Create class MathOperations.
2. Define method multiply(*args).

Sumit Sir Notes
3. Initialize the answer as 1.
4. Loop through all numbers.
5. Multiply each number.
6. Return the final product.'''
class MathOperations: 
    def multiply(self,*args):
        ans = 1
        for num in args:
            ans *= num
        return ans 

A = MathOperations()
print(A.multiply(5,5,5))
       
    
```

```python
# 5. Create a Display class that prints different messages based on the argument type.
'''Concept
Method Overloading + Type Checking
What to do
Accept:
• Integer
• String
• List
How to solve
1. Create class Display.
2. Create method show().
3. Use isinstance().
4. Print different messages depending on the object's type.'''


class Display:
    def show(self, value):
        if isinstance(value, int):
            print("Integer.")
        elif isinstance(value, str):
            print("String.")
        elif isinstance(value, list):
            print(" List.")
        else:
            print("Unknown data type.")



d = Display()


d.show(10)
d.show("Hello")
d.show([1, 2, 3])
d.show(10.10)
```

```python
#7. Create a PrintData class that prints one or many values.
'''Concept
Method Overloading using *args
What to do
Print:
• One value
• Multiple values
How to solve
1. Create class PrintData.
2. Create method print_data(*args).
3. Loop through the arguments.
4. Print each value.'''

class PrintData:
    def print_data(self,*args):
        for data in args:
            print(data)

A = PrintData()

print(A.print_data("Hello"))
A.print_data("Python", 100, [1, 2, 3])
    
```

```python
#10. Create an Employee class and override calculate_salary() in Manager.
'''Concept
Method Overriding
What to do
• Employee → Normal salary
• Manager → Salary + Bonus
How to solve
1. Create parent class.
2. Define calculate_salary().
3. Inherit Manager.
4. Override the method.
5. Print the manager's salary.'''

class Employee:
    def calculate_salary(self):
        print("Normal salary 50000")

class Manager(Employee):
    def calculate_salary(self):
        Salary = 50000
        Bonus = 10000
        print(f"manager's salary is {Salary + Bonus}")
        

B = Manager()

B.calculate_salary()
```

```python
#11. Create Shape, Circle, and Rectangle classes.
'''Concept
Method Overriding
What to do
Override the area() method.
How to solve
1. Create Shape.
2. Add area().
3. Inherit Circle.
4. Inherit Rectangle.
5. Override area() in each child class.'''

class Shape:
    def area(self):
        print("area of the Shape is...")

class Circle(Shape):
    def area(self,r):
        area = (22/7)*r*r
        print(f"area of the Circle is...:{area}")

class Rectangle(Shape):
    def area(self,l,b,h):
        area = l*b*h
        print(f"area of the Rectangle is...:{area}")

A = Circle()
B = Rectangle()

A.area(2)
B.area(2,3,4)
```

```python
#16. Create a Book class and overload the + operator.
'''Concept
Operator Overloading
What to do
Add the number of pages of two books.
How to solve
1. Create class Book.
2. Store pages.
3. Overload __add__().
4. Return the total pages.'''

class Book:
    def __add__(self,pages):
        self.pages = pages

    def __add__(self, other):
        return self.pages + other.pages


book1 = Book()
book2 = Book()

book1 = 250
book2 = 100
total_pages = book1 + book2

print("Total pages:", total_pages)
        
```

```python
#18. Create a Rectangle class and overload the == operator.

'''Sumit Sir Notes
Concept
Equality Operator Overloading (__eq__)
What to do
Compare rectangles using their area.
How to solve
1. Create class Rectangle.
2. Store length and width.
3. Calculate area.
4. Overload __eq__().
5. Compare the areas.'''

class Rectangle:
    def __init__(self,l,b):
        self.l = l
        self.b = b
    def __eq__(self,other):
        ar1 = self.l * self.b
        ar2 = other.l * other.b
        return ar1 == ar2


        

a = Rectangle(2,3)
b = Rectangle(2,5)
if a == b:
    print("Both rectangles have the same area.")
else:
    print("Rectangles have different areas.")
```

```python
#19. Create a Money class and overload the + and - operators.
'''Concept
Multiple Operator Overloading
What to do
Perform addition and subtraction of money amounts.
How to solve
1. Create class Money.
2. Store the amount.
3. Overload __add__().
4. Overload __sub__().
5. Return new Money objects after each operation.'''

class Money:
    def __init__(self,a):
        self.a = a 

    def __add__(self,other):
        return self.a + other.a

    def __sub__(self,other):
        return self.a - other.a

m1 = Money(9)
m2 = Money(7)


sum1 = m1 + m2
sub = m1 - m2


print(sum1)
print(sub)
```

```python
# 20. Create a Distance class and overload <, >, and ==.
'''Concept
Comparison Operator Overloading
What to do
Compare two distance objects.

Sumit Sir Notes
How to solve
1. Create class Distance.
2. Store the distance value.
3. Overload:
o __lt__() for <
o __gt__() for >
o __eq__() for ==
4. Compare the stored distance values.
5. Return True or False.'''

class Distance:
    def __init__(self, distance):
        self.distance = distance

    
    def __lt__(self, other):
        return self.distance < other.distance

    
    def __gt__(self, other):
        return self.distance > other.distance

   
    def __eq__(self, other):
        return self.distance == other.distance



d1 = Distance(120)
d2 = Distance(120)

print("Distance 1:", d1.distance)
print("Distance 2:", d2.distance)

print(d1 < d2)
print(d1 > d2)
print(d1 == d2) 
```
