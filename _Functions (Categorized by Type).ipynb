```python
#13. Write a function full_name(first, last) and call it using positional arguments. 
def full_name(first, last):
    print(first)
    print(last)


#fuction calls
full_name("Seelam","Reddy")
```

```python
#14. Call the same function using keyword arguments. 
def keyword(name,age,mob_no):
    return f"{name}  {age}  {mob_no}"


#fuction calls
keyword("anil",10,123456)
```

```python
#16. Create a function calculate(a, b, operation) and show how keyword args improve readability. 
def calculate(a, b, operation):
    if operation == "+":
        print(a+b)
    elif operation =="-":
        print(a-b)
    elif operation == "*":
        print(a*b)
    elif operation == "/":
        print(a/b)
        
    
    else:
        
        print("invaild operation")

#fuction calls

calculate(12,23,"+")
```

```python
#21. Write a function find_max(*args) that returns the maximum value. 
def find_max(*args):
    print(max(args))


#fuction calls
find_max(10,20,30)
```

```python
#22. Make a function concatenate(*args) that joins all string arguments with space. 
def concatenate(*args):
   
    result = ""
    for item in args:
        result += str(item)
    return result

# Function call
concatenate("Hello", "       ", "World", "!")
```

```python
#23. Write a function multiply(*args) that multiplies all numbers. def multiply(*num):
def multiply(*args):
    if not args:
        return 0

    result = 1
    for num in args:
        result *= num
    return result

# Function calls
print(multiply())          
print(multiply(2, 3, 4))   
print(multiply(5, 10))
```
