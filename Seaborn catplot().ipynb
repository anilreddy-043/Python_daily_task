# Seaborn catplot() — 30 Questions

```python
# A. Titanic Dataset — 15 Questions
#Use:
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
titanic = sns.load_dataset("titanic")
titanic
```

```python
#1. Survival by Gender
#Use catplot() to show the number of passengers who survived and did not survive for each gender.
#• X-axis → sex
#• Hue → survived
#• Chart → count
sns.catplot(data = titanic,
           x = "sex",
            hue = "survived",
           kind = "count")
plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
# 2. Survival by Passenger Class
# Create a categorical plot showing the number of passengers in each class. Use:
# • x="class"
# • kind="count"
sns.catplot(data = titanic,
             x="class",
            kind="count"
           )
plt.title("Survival by Passenger Class")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
# 3. Survival Count by Class
# Show the number of survivors and non-survivors for each passenger class.
#Use:
# x="class"
# hue="survived"
# kind="count"
sns.catplot(data = titanic,
            x="class",
            hue="survived",
            kind="count"
           )
plt.title("Survival Count by  Class")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
# 4. Survival Percentage by Gender
# Create a categorical visualization comparing survival across sex.
# Use catplot() and choose an appropriate categorical plot.
# Then identify which gender has a higher survival rate.

sns.catplot(
    data=titanic,
    x="sex",
    y="survived",
    kind="bar",
    height=5,
    aspect=0.5
)

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")
plt.show()
```

```python
# 5. Age Distribution by Gender
# Use catplot() to compare passenger age across different genders.
# Try an appropriate kind for showing the distribution.
sns.catplot(data = titanic,
           x = "sex",
           y = "age",
            kind = "bar"
           )
```

```python
# 6. Age Distribution by Passenger Class
# Create a categorical plot showing age distribution for:
#• First class
#• Second class
# • Third class
sns.catplot(data =titanic,
           x = "class",
           y = "age")
```

```python
# 7. Fare by Passenger Class
#Use catplot() to compare fare across the three passenger classes.
#Add:
#• Appropriate color
# • Title
# • X/Y labels
sns.catplot(data = titanic,
           x= "pclass",
           y = "fare",
            kind = "bar",
           palette = "dark")

plt.title("Fare by Passenger Class")
plt.xlabel("Fare")
plt.ylabel("Passenger Class")
plt.show()
```

```python
# 8. Fare Distribution by Gender
# Create a categorical visualization of fare for male and female passengers.
#Identify whether the fare distributions appear different.
sns.catplot(
    data=titanic,
    x="sex",
    y="fare",
    kind="box",
    height=5,
    aspect=1
)

plt.title("Fare Distribution by Gender")
plt.show()
```

```python
# 9. Age vs Passenger Class
# Use a categorical plot to compare the distribution of age across class.
# Add sex using hue.
sns.catplot(data = titanic,
           x = "pclass",
           y = "age",
           hue = "sex")
plt.show()
```

```python
#10. Survival by Class and Gender
#Create a categorical visualization showing survival for each passenger class and gender.
# Use:
# x="class"
# hue="sex"
#Use an appropriate kind.
sns.catplot(data = titanic,
            x= "class",
            hue = "sex",
            kind = "count")
```

```python
# 11. Embark Town vs Survival
# Create a count plot using catplot() showing passenger counts for each embark_town.
# Use survived as hue.
sns.catplot(data = titanic,
           x="embark_town",
           hue = "sex",
           kind = "count")
```

```python
# 12. Passenger Class vs Fare with Gender
# Create a categorical visualization comparing fare across passenger classes.
#Use:
#x="class"
# y="fare"
#hue="sex"
# Choose an appropriate kind.
sns.catplot(data = titanic,
            x="class",
            y="fare",
            hue="sex",
              kind = "violin")
```

```python
#13. Survival vs Age Group
#Create an age_group column:
#• Child → Age < 18
#• Adult → 18–59
#• Senior → 60+
#Use catplot() to compare survival across these groups.

titanic["age_group"] = np.where(
    titanic["age"] < 18, "Child",
    np.where(titanic["age"] < 60, "Adult", "Senior")
)

sns.catplot(
    data=titanic,
    x="age_group",
    y="survived",
    kind="bar",
    height=5,
    aspect=1.2
)

plt.title("Survival Rate by Age Group")
plt.ylabel("Survival Rate")
plt.xlabel("Age Group")
plt.show()
```

```python
#14. Class, Gender and Survival
#Create a categorical visualization where:
#• X → class
#• Y → survival-related measure
#• Hue → sex
#• Columns → survived
#Use col= to create separate panels.

sns.catplot(
    data=titanic,
    x="pclass",
    y="survived",
    hue="sex",
    col="survived",
    kind="bar"

)

plt.show()
```

```python
# 15. Titanic Multi-Panel Analysis
#Create a catplot() with:
#• X → class
#• Y → fare
#• Hue → sex
#• Columns → survived
#Use an appropriate categorical kind.
#Add:
#• Figure size
#• Title
#• Color palette
#• Appropriate labels
#Explain what you observe.
sns.catplot(
    data=titanic,
    x = "pclass",
    y = "fare",
    hue = "sex",
    col = "survived",
    kind = "bar")
plt.title("Titanic Multi-Panel Analysis")
plt.show()
```

# B. Tips Dataset — 15 Questions

```python
###### Load the dataset:
tips = sns.load_dataset("tips")
#The dataset contains variables such as:
#• total_bill
#• tip
#• sex
#• smoker
#• day
#• time
#• size
tips
```

```python
#16. Total Bill by Day
#Use catplot() to compare total_bill across:
#• Thur
#• Fri
#• Sat
#• Sun
#Choose an appropriate kind.

sns.barplot(
    data=tips,
    x = "day",
    y = "total_bill")
plt.title("Total Bill by Day")
plt.show()
```

```python
#17. Tip by Day
#Create a categorical visualization showing tip distribution across different days.

sns.barplot(
    data=tips,
    x = "day",
    y = "tip")
plt.title(" Tip by Day")
plt.show()
```

```python
#18. Total Bill by Gender
#Compare total_bill between:
#• Male
#• Female
#Use catplot().
sns.catplot(data = tips,
           x = "sex",
           y = "total_bill")
plt.title("Total Bill by Gender")
plt.show()
```

```python
#19. Tip by Gender
#Create a categorical plot comparing tips given by male and female customers.
sns.catplot(data = tips,
           x = "sex",
           y = "tip")
plt.title("Tip by Gender")
plt.show()
```

```python
#20. Total Bill by Smoker Status
#Compare the distribution of total_bill between:
#• Smokers
#• Non-smokers
#Use hue="smoker" where appropriate.
sns.boxplot(data = tips,
           x = "smoker",
           y = "total_bill")
plt.title("Total Bill by Smoker Status")
plt.show()
```

```python
#21. Tip by Smoker Status
#Create a categorical visualization comparing tips between smokers and non-smokers.
sns.catplot(data = tips,
           x = "smoker",
           y = "total_bill")
plt.title("Total Bill by Smoker Status")
plt.show()
```

```python
#22. Total Bill by Day and Gender
#Create a plot with:
#x="day"
#y="total_bill"
#hue="sex"
#Choose an appropriate categorical kind.

sns.catplot(data = tips,
           x = "day",
           y = "total_bill",
           hue = "sex")
plt.title("Total Bill by Day and Gender")
plt.show()
```

```python
#23. Tip by Day and Gender
#Compare tips across different days and genders.
#Use:
#x="day"
#y="tip"
#hue="sex"

sns.boxenplot(data = tips,
           x = "day",
           y = "tip",
           hue="sex")
plt.title("Tip by Day and Gender")
plt.show()
```

```python
#24. Total Bill by Time
#Compare lunch and dinner bills.
#Use:
#x="time"
#y="total_bill"
#Choose an appropriate kind.

sns.catplot(data = tips,
           x = "time",
           y = "total_bill",
           #hue="sex"
           )
plt.title("Total Bill by Time")
plt.show()
```

```python
#26. Bill by Day and Smoker Status
#Create a categorical plot:
#x="day"
#y="total_bill"
#hue="smoker"
#Determine whether smokers and non-smokers show different bill distributions.

sns.boxenplot(data = tips,
           x = "day",
           y = "total_bill",
            hue ="smoker"
           )
plt.title("Bill by Day and Smoker Status")
plt.show()
```

```python
#27. Tip by Day and Time
#Create a categorical visualization with:
#x="day"
#y="tip"
##hue="time"
#Compare the tip distributions.

  
sns.barplot(data = tips,
           x = "day",
           y = "tip",
            hue ="time"
           )
plt.title("Tip by Day and Time")
plt.show()  
```

```python
#28. Party Size Analysis
#Use size as a categorical variable.
#Create a categorical plot comparing total_bill for different party sizes.
#For example:
#x="size"
#y="total_bill"

  
sns.barplot(data = tips,
           x = "size",
           y = "total_bill"
           )
plt.title("Party Size Analysis")
plt.show()  
```

```python
#29. Multi-Panel Tips Analysis
#Create a catplot() with:
#• X → day
#• Y → total_bill
#• Hue → sex
#• Columns → time
#Choose an appropriate kind.
#Then compare lunch vs dinner.
  
sns.catplot(data = tips,
           x = "day",
           y = "total_bill",
            hue = "sex",
            col = "time"
           )
plt.title("Multi-Panel Tips Analysis")
plt.show() 
```

```python
#30. Final Seaborn catplot() Challenge
##Using the Tips dataset, create a professional categorical visualization with:
#x="day"
#y="total_bill"
#hue="sex"
#col="time"
#Choose the most appropriate kind.
#Customize:
#• palette
##• height
#• aspect
#• col_wrap if appropriate
#• Title
##• Axis labels
#• Legend

sns.catplot(data = tips,
           x = "day",
           y = "total_bill",
            hue = "sex",
            col = "time",
            kind = "boxen",
            palette = "Set2",
            height = 5,
            aspect = 1.2
            
            
           )
plt.title("Multi-Panel Tips Analysis")
plt.legend(["lunch vs dinner"])
plt.xlabel("days")
plt.ylabel("total_bills")
plt.show() 
```
