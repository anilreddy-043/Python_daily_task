```python
#1. Class Name: Shape 
#Problem Create an abstract class named Shape. 
#Explanation A shape can have different formulas for area and perimeter. 
#Therefore, create two abstract methods:
#• area()
#• perimeter() 
#Create subclasses like Circle and Rectangle and implement both methods. 

from abc import ABC, abstractmethod
import math

# Abstract class
class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


# Circle subclass
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius


# Rectangle subclass
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


# Driver code
circle = Circle(5)
rectangle = Rectangle(4, 6)

print("Circle Area:", circle.area())
print("Circle Perimeter:", circle.perimeter())

print("Rectangle Area:", rectangle.area())
print("Rectangle Perimeter:", rectangle.perimeter())
```

```python
#2. Class Name: Vehicle
#Problem Create an abstract class named Vehicle. 
#Explanation Every vehicle can start and stop, but each vehicle does it differently. 
#Create two abstract methods: 
#• start() 
# • stop() 
#Create a Car class that implements both methods.


from abc import ABC, abstractmethod
import math

# abstract class
class vehicle(ABC):
    
    @abstractmethod
    def start(self):
        pass
        
    @abstractmethod
    def stop(self):
        pass

        
# car sub class
 
class car(vehicle):
    def start(self):
        print("car started")

    def stop(self):
        print("car stopped")

# Driver code
car = car()
car.start()
car.stop()
```

```python
#4. Class Name: Payment 
#Problem Create an abstract class named Payment. 
#Explanation Every payment processes money differently.
#Create an abstract method: 
#• process_payment(amount)  
#Create subclasses: 
#• CreditCard  
#• Cash 
#Each should process the payment in its own way.

from abc import ABC, abstractmethod

class payment(ABC):
    
    @abstractmethod
    def process_payment(self,amount):
        pass


        
#CreditCard subclasses

class CreditCard(payment):
    
    def process_payment(self,amount):
        print(f"Processing credit card payment of ₹{amount}")

class Cash(payment):
    def process_payment(self,amount):
        print(f"Processing cash payment of ₹{amount}")

# Driver code

CreditCard = CreditCard()
Cash = Cash()

CreditCard.process_payment(10000000)
Cash.process_payment(50000)
```
